# 03. 선형결합 · 생성(span) · 일차독립

## 1. 선형결합 (Linear Combination)

벡터들을 스칼라배 해서 더한 것: c1*v1 + c2*v2

계수 c를 바꿔가며 새로운 벡터를 만들어낸다.

```python
import numpy as np 

v1 = np.array([1, 0])
v2 = np.array([0, 1])

def lin_comb(c1, c2):
    return c1 * v1 + c2 * v2

print("3*v1 + 2*v2 =", lin_comb(3, 2))
print("-1*v1 + 5*v2 =", lin_comb(-1, 5))
print("0*v1 + 0*v2 =", lin_comb(0, 0))

# 계수를 바꾸면 평면의 어떤 점이든 만들 수 있다. 
# 목표: [7, -3]을 만들려면 계수가 뭐여야 할까? 
print("목표 [7, -3] =", lin_comb(7, -3))
```

## 2. 생성 (Span)

선형결합으로 만들 수 있는 모든 벡터의 집합.

- 방향이 다른 두 벡터 → 평면 전체를 채움 (span = 2D)
- 나란한 두 벡터 → 직선 하나만 (span = 1D)

```python
import numpy as np

# 경우 A: 방향이 다른 두 벡터
v1 = np.array([1, 0])
v2 = np.array([0, 1])

# 경우 B: 나란한 두 벡터 (w2 = w1의 2배)
w1 = np.array([1, 2])
w2 = np.array([2, 4])

# '방향이 같은지'를 어떻게 알까? → 두 벡터를 행으로 쌓은 행렬의 'rank'
# rank = 서로 독립인 (방향이 다른) 벡터의 개수
A = np.array([v1, v2])
B = np.array([w1, w2])

print("경우 A rank =", np.linalg.matrix_rank(A))
print("경우 B rank =", np.linalg.matrix_rank(B))
```

## 3. 일차독립 (Linear Independence)

어떤 벡터도 나머지의 선형결합으로 만들 수 없으면 '독립'.

- rank == 벡터 개수 → 일차독립 (모두 고유한 방향)
- rank < 벡터 개수 → 일차종속 (군더더기 존재)

```python
import numpy as np

def is_independent(*vectors):
    M = np.array(vectors)
    rank = np.linalg.matrix_rank(M)
    n = len(vectors)
    print(f" rank={rank}, 벡터개수={n} →", "독립" if rank == n else "종속")
    return rank == n

print("[1,0],[0,1]:", is_independent([1,0],[0,1]))
print("[1,2],[2,4]:", is_independent([1,2],[2,4]))
print("[1,0],[0,1],[1,1]:", is_independent([1,0],[0,1], [1,1]))
```
