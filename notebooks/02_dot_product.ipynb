# 02. 내적 (Dot Product) — 각도와 투영

## 1. 내적과 각도

<!-- 내적은 v 길이 x w 길이 x 사이각의 cos값 -->
v · w = ‖v‖‖w‖cosθ  →  cosθ = (v·w) / (‖v‖‖w‖)

```python
import numpy as np

def angle_between(v, w):
    """두 벡터 사이 각도를 '도(degree)' 단위로 반환"""
    cos_theta = np.dot(v, w) / (np.linalg.norm(v) * np.linalg.norm(w))
    cos_theta = np.clip(cos_theta, -1.0, 1.0)
    return np.degrees(np.arccos(cos_theta))

# 세 가지 경우 확인
a = np.array([1, 0])
b = np.array([1, 1]) # 45도
c = np.array([0, 1]) # 90도
d = np.array([-1, 0]) # 180도

# NOTE: 무리수는 정확히 담을 수 없음 
# 1e-14 수준 오차 발생
print("a·b =", np.dot(a, b), "| 각도 =", angle_between(a, b), "도")
print("a·c =", np.dot(a, c), "| 각도 =", angle_between(a, c), "도")
print("a·d =", np.dot(a, d), "| 각도 =", angle_between(a, d), "도")
```

## 2. 투영 (Projection)

v를 w 위에 떨어뜨린 그림자.

proj_w(v) = (v·w / w·w) * w

v = (w 방향 성분) + (w에 수직인 나머지) 로 분해된다.

```python
import numpy as np 

def project(v, w):
    """v를 w 위로 투영한 벡터 변환"""
    scale = np.dot(v, w) / np.dot(w, w)
    return scale * w

v = np.array([2, 3])
w = np.array([4, 0])

p = project(v, w)
perp = v - p

print("투영 proj_w(v) =", p)
print("수직 성분 (v - p) =", perp)
print("p + perp =", p + perp, " (원래 v와 같아야 함")
print("perp · w =", np.dot(perp, w), " (0이면 수직 확인!)")
```

## 3. 코사인 유사도 (Cosine Similarity)

cosθ = (v·w) / (‖v‖‖w‖)  를 '유사도 점수'로 사용 (-1 ~ 1).

크기는 무시하고 '방향'만 비교한다 → 추천/검색/문서 비교의 핵심

```python
import numpy as np

def cosine_similarity(v, w):
    return np.dot(v, w) / (np.linalg.norm(v) * np.linalg.norm(w))

# 예: 세 사람의 영화 취향 [액션, 로맨스, 코미디] 점수
me = np.array([5,1,4])
alice = np.array([4,1,5])
bob = np.array([1,5,1])

print("나 vs Alice:", cosine_similarity(me, alice))
print("나 vs bob  :", cosine_similarity(me, bob))

# 크기 무시 확인: alice 점수를 10배 키워도 유사도는 그대로?
print("나 vs Alice * 10:", cosine_similarity(me, alice * 10))
```

## 정리

- 도(degree), 라디안(radian): 둘 다 '각도'를 재는 단위 (서로 변환 가능)
  - 도: 0 ~ 360
  - 라디안: 0 ~ 2π
- 코사인 값: 각도가 아니라 -1 ~ 1 사이의 '비율'

cosθ = (v·w) / (‖v‖ × ‖w‖)
- cosθ = -1 : 정반대 방향 (역의 관계)
- cosθ =  0 : 직교 (관계 없음)
- cosθ =  1 : 같은 방향 (정의 관계)
