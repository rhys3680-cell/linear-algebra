# 05. 연립방정식 풀이 (Ax = b)

## 1. Ax = b 풀기

연립방정식을 A(계수행렬), x(미지수), b(우변)로 묶으면 Ax = b.

"A 변환을 거쳐 b에 도착하는 입력 x는?" → x = A⁻¹b (A를 되돌리기)."
np.linalg.solve(A, b)

```python
import numpy as np

# 2x + y = 5
#  x + 3y = 10
A = np.array([[2, 1],
              [1, 3]])
b = np.array([5, 10])

# 방법 1: 역행렬로 (개념용)
x_inv = np.linalg.inv(A) @ b

# 방법 2: solve (실전 표준)
x_solve = np.linalg.solve(A, b)

print("역행렬로 x =", x_inv)
print("solve로  x =", x_solve)

# 검증: 구한 x를 다시 넣으면 b가 나오나?
print("A @ x =", A @ x_solve, " (b와 같아야 함:", b, ")")
```

## 2. 해가 없거나 무한한 경우 (det = 0)

x = A⁻¹b 인데, det(A)=0 이면 역행렬이 없다 → 깔끔한 해가 없다.

- 해 무한: 두 식이 사실상 같음 (직선이 겹침)
- 해 없음: 두 식이 모순 (평행선)
둘 다 det=0. solve는 이 때 에러를 낸다.

```python
import numpy as np

# 경우 1: 해 무한 (2번째 = 1번째 x2, 우변도 x2)
A1 = np.array([[1, 2], [2, 4]])
b1 = np.array([4, 8])

# 경우 2: 해 없음 (좌변은 x2인데 우변은 모순)
A2 = np.array([[1, 2], [2, 4]])
b2 = np.array([4, 9])

print("A1 det =", np.linalg.det(A1)) # 0
print("A2 det =", np.linalg.det(A2)) # 0 (같은 A)
print()

# det=0이면 solve가 어떻게 되나?
for name, A, b in [("1 해 무한", A1, b1), ("2 해 없음", A2, b2)]:
    try:
        x = np.linalg.solve(A, b)
        print(name, "solve → ", x)
    except np.linalg.LinAlgError as e:
        print(name, "solve → 에러:", e)
```

## 3. rank로 해의 종류 판별

A의 rank와 [A|b](첨가행렬)의 rank 를 비교한다.

- rank(A) = rank([A|b]) = 미지수개수 → 해 1개
- rank(A) = rank([A|b]) < 미지수개수 → 해 무한
- rank(A) < rank([A|b])               → 해 없음 (b가 도달 불가)

```python
import numpy as np

def classify(A, b):
    rank_A = np.linalg.matrix_rank(A)
    Ab = np.column_stack([A, b]) # A 옆에 b를 한 칸 붙임 (첨가행렬)
    rank_Ab = np.linalg.matrix_rank(Ab)
    n = A.shape[1]

    if rank_A < rank_Ab:
        kind = "해 없음 (모순)"
    elif rank_A == n:
        kind = "해 1개 (유일)"
    else:
        kind = "해 무한"
    print(f" rank(A)={rank_A}, rank([A|b])={rank_Ab}, 미지수={n} → {kind}")

print("정상 (한 점 교차):")
classify(np.array([[2,1],[1,3]]), np.array([5,10]))

print("1 해 무한 (겹친 직선):")
classify(np.array([[1,2],[2,4]]), np.array([4,8]))

print("2 해 없음 (평행/모순):")
classify(np.array([[1,2],[2,4]]), np.array([4,9]))
```
