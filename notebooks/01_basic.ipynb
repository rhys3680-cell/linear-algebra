# 01. 벡터 기초

벡터(vector)는 **방향과 크기를 가진 수의 묶음**입니다. 예: [1, 2, 3]

```python
import numpy as np 

# 벡터 만들기
v = np.array([1, 2, 3])
w = np.array([4, 5, 6])

print("v =", v)
print("w = ", w)
```

## 1. 벡터의 덧셈과 뺄셈

같은 위치의 원소끼리 더하고 뺍니다. (원소별 연산)

```python
print("v + w =", v + w) 
print("v - w =", v - w)
```

```python
# 직접 풀어보기: a = [2, 0, -1], b = [1, 1, 1]을 만들고
# a + b와 a - b를 출력해보세요.

a = np.array([2, 0, -1])
b = np.array([1, 1, 1])

print("a + b =", a + b)
print("a - b =", a - b)
```

## 2. 스칼라 곱

벡터에 숫자(스칼라)를 곱하면 모든 원소가 그만큼 커집니다. → 크기만 변하고 방향은 그대로

```python
print("2 * v =", 2 * v)
print("0.5 * v =", 0.5 * v)
```

```python
# 직접 풀어보기: v 에 -3 을 곱한 결과를 출력해보세요.

print("3 * v =", 3 * v)
```

## 3. 내적 (dot product)

두 벡터를 원소별로 곱한 뒤 모든 더한 **하나의 숫자**입니다.

v · w = 1x4 + 2x5 + 3x6 = 32

→ 두 벡터가 얼마나 "같은 방향"인지를 나타냅니다.

```python
print("np.dot(v, w) =", np.dot(v, w))
print("v @ w =", v @ w)
```

```python
# 직접 풀어보기: a = [1, 0], b = [0, 1]의 내적을 구해보세요.
# 결과가 0이면 두 벡터는 '수직(직교)'입니다. 확인해보세요!

a = np.array([1, 0])
b = np.array([0, 1])

print("np.dot(a, b) =", np.dot(a, b))
print("a @ b =", a @ b)
```

## 4. 벡터의 크기 (노름, norm)

벡터의 "길이"입니다. 피타고라스 정리의 확장이에요.

‖v‖ = √(1² + 2² + 3²) = √14 ≈ 3.74

```python
print("np.linalg.norm(v) =", np.linalg.norm(v)) # 3.7416...

## 직접 계산으로 확인
print("직접 계산 =", np.sqrt(np.sum(v**2)))
```

```python
# 직접 풀어보기: c = [3, 4] 의 크기를 구해보세요.
c = np.array([3, 4])

print("c의 길이는 ", np.linalg.norm(c))
```
