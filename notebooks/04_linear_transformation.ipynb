# 04. 행렬 = 선형변환

## 1. 변환 행렬 직접 만들기

규칙: 행렬의 1열 = î=[1,0]가 가는 곳, 2열 = ĵ=[0,1]가 가는 곳.

이 규칙으로 회전/확대/반사를 외우지 않고 설계한다.

```python
import numpy as np 

# 규칙: [1열=î 행선지, 2열=ĵ 행선지] 를 '열로 세워' 적는다
scale2 = np.array([[2, 0], [0, 2]])
rotate90 = np.array([[0, -1], [1, 0]])
flip_x = np.array([[1, 0], [0, -1]])

p = np.array([2, 3])

print("원래 점 :", p)
print("2배 확대 :", scale2 @ p)
print("90도 회전 :", rotate90 @ p)
print("x축 반사 :", flip_x @ p)
```

```python
# 직접 확인: 각 행렬의 '열'이 정말 î, ĵ의 행선지인가?
print("\nrotate90의 1열 (î가 가는 곳):", rotate90[:, 0])
print("\nrotate90의 2열 (ĵ가 가는 곳):", rotate90[:, 1])
```

## 2. 변환 합치기 (행렬곱)

두 변환을 연달아 = 두 행렬의 곱. 단, 오른쪽부터 적용된다.

A @ B @ p = "B 먼저, 그 다음 A". 그리고 순서 바꾸면 결과가 달라진다 (AB ≠ BA).


```python
import numpy as np

rotate90 = np.array([[0, -1], [1, 0]])
flip_x = np.array([[1, 0], [0, -1]])

p = np.array([2, 3])

rot_then_flip = flip_x @ rotate90
flip_then_rot = rotate90 @ flip_x

print("회전→반사 한 행렬:\n", rot_then_flip)
print("반사→회전 한 행렬:\n", flip_then_rot)
print()

# 한 번에 적용 vs 단계별 적용 — 같은지 확인
print("flip_x @ (rotate90 @ p) =", flip_x @ (rotate90 @ p))   # 단계별
print("(flip_x @ rotate90) @ p =", rot_then_flip @ p)         # 합쳐서 한번에
print()

# 순서를 바꾸면 결과가 다른가?
print("회전→반사 결과:", rot_then_flip @ p)
print("반사→회전 결과:", flip_then_rot @ p, "  ← 다르다! (AB ≠ BA)")
```

## 3. 행렬식 (Determinant)

det = 변환이 넓이를 몇 배로 바꾸는가.

- det = k → 넓이 k배
- det = 0 → 공간을 찌그러지게 만듦 (직선/점).
- det < 0 → 넓이 |det|배 + 뒤집힘(반사)

```python
import numpy as np

scale2 = np.array([[2, 0], [0, 2]])
rotate90 = np.array([[0, -1], [1, 0]])
shear = np.array([[1, 1], [0, 1]])
flip_x = np.array([[1, 0], [0, -1]])
squash = np.array([[1, 2], [2, 4]])

for name, M in [("2배 확대", scale2), ("90도 회전", rotate90),
                ("전단", shear), ("x축 반전", flip_x), ("찌그러짐", squash)]:
    d = np.linalg.det(M)
    print(f"{name:8s} det = {d:5.1f}")

# det=0이면 정말 역행렬이 없는가? 
print("\nsquash의 rank =", np.linalg.matrix_rank(squash))
try:
    np.linalg.inv(squash)
except np.linalg.LinAlgError as e:
    print("squash 역행렬 시도 →", e)
```
