# 행렬 = 선형변환을 '격자 변형'으로 기하학적으로 표현.

행렬의 열 = 기본 벡터 î, ĵ의 행선지.
정사각 격자가 행렬을 거치면 평행사변형 격자로 변형된다.

```python
import numpy as np
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

def draw_grid(ax, A, title):
    """행렬 A로 변형된 격자 + î, ĵ 화살표를 그린다."""
    lim = 3
    lines = np.arange(-lim, lim + 1)

    # 격자선: 변형 전(회색) vs 변형 후(파랑)
    for c in lines: 
        # 세로선, 가로선의 양 끝점을 A로 변환
        for pts in ([[c, -lim], [c, lim]], [[-lim, c], [lim, c]]):
            p = np.array(pts).T
            tp = A @ p
            ax.plot(p[0], p[1], color="lightgray", lw=0.8, zorder=1)
            ax.plot(tp[0], tp[1], color="royalblue", lw=1.0, zorder=2)

    # î, ĵ의 행선지 = A의 열
    i_hat, j_hat = A[:, 0], A[:, 1]
    ax.quiver(0, 0, i_hat[0], i_hat[1], angles="xy", scale_units="xy",
              scale=1, color="crimson", zorder=5, width=0.015)
    ax.quiver(0, 0, j_hat[0], j_hat[1], angles="xy", scale_units="xy",
              scale=1, color="green", zorder=5, width=0.015)

    ax.set_xlim(-lim, lim)
    ax.set_ylim(-lim, lim)
    ax.set_aspect("equal")
    ax.axhline(0, color="gray", lw=0.6)
    ax.axvline(0, color="gray", lw=0.6)
    det = np.linalg.det(A)
    ax.set_title(f"{title}\ndet = {det:.0f} (넓이 배율)")


fig, axes = plt.subplots(1, 3, figsize=(15, 5))

draw_grid(axes[0], np.array([[2, 0], [0, 2]]), "2배 확대")
draw_grid(axes[1], np.array([[0, -1], [1, 0]]), "90도 회전")
draw_grid(axes[2], np.array([[1, 1], [0, 1]]), "전단(shear) - 비스듬히 밀기")

# 범례
axes[0].plot([], [], color="lightgray", label="변형 전 격자")
axes[0].plot([], [], color="royalblue", label="변형 후 격자")
axes[0].plot([], [], color="crimson", label="i (1열)")   # quiver → plot
axes[0].plot([], [], color="green", label="j (2열)")     # quiver → plot
axes[0].legend(loc="upper left", fontsize=8)


plt.tight_layout()
plt.show()
```
