# 벡터의 길이(norm) 시각화

- 2D: [3, 4] → 길이 5 (피타고라스로 직관적 확인)
- 3D: [1, 2, 3] → 길이 sqrt(14) ≈ 3.74
- 벡터는 기하학적 길이의 개념이 성립
- 행렬부터는 거리의 개념만 존재

```python
import numpy as np 
import matplotlib.pyplot as plt

plt.rcParams["font.family"] = "Malgun Gothic"
plt.rcParams["axes.unicode_minus"] = False

fig = plt.figure(figsize=(12,5))

# ---- 2D 벡터 [3, 4] ----
ax1 = fig.add_subplot(1, 2, 1)
v2 = np.array([3, 4])
length2 = np.linalg.norm(v2)

# 벡터 화살표
ax1.quiver(0, 0, v2[0], v2[1], angles="xy", scale_units="xy", scale=1,
           color="crimson", width=0.012)
# 직각삼각형 보조선 (밑변 3, 높이 4)
ax1.plot([0, 3], [0, 0], "k--", alpha=0.5)
ax1.plot([3, 3], [0, 4], "k--", alpha=0.5)
ax1.text(1.5, -0.4, "밑변 = 3", ha="center")
ax1.text(3.2, 2, "높이 = 4", va="center")
ax1.text(1.2, 2.2, f"길이 = {length2:.0f}", color="crimson", fontsize=13,
         rotation=53)

ax1.set_xlim(-1, 5)
ax1.set_ylim(-1, 5)
ax1.set_aspect("equal")
ax1.grid(True, alpha=0.3)
ax1.axhline(0, color="gray", lw=0.8)
ax1.axvline(0, color="gray", lw=0.8)
ax1.set_title(f"2D 벡터 [3, 4]  ·  길이 = √(3²+4²) = {length2:.0f}")

# ---- 3D 벡터 [1, 2, 3] ----   ← 빠져있던 부분
ax2 = fig.add_subplot(1, 2, 2, projection="3d")
v3 = np.array([1, 2, 3])
length3 = np.linalg.norm(v3)

ax2.quiver(0, 0, 0, v3[0], v3[1], v3[2], color="royalblue", arrow_length_ratio=0.1)
ax2.scatter(*v3, color="royalblue", s=40)
ax2.text(v3[0], v3[1], v3[2] + 0.2, f"[1,2,3]  길이={length3:.2f}", color="royalblue")

ax2.set_xlim(0, 3); ax2.set_ylim(0, 3); ax2.set_zlim(0, 3)
ax2.set_xlabel("x"); ax2.set_ylabel("y"); ax2.set_zlabel("z")
ax2.set_title(f"3D 벡터 [1, 2, 3]  ·  길이 = sqrt(14) = {length3:.2f}")

plt.tight_layout()
plt.show()
```
